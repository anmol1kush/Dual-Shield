```python
import base64
from datasets import concatenate_datasets
```

```python
def encode_base64(text):
    return base64.b64encode(text.encode("utf-8")).decode("utf-8")
```

```python
def create_input_shield(example):
    encoded = encode_base64(example["adversarial"])

    return {
        "input_text": f"Summarize:\n<DATA>{encoded}</DATA>",
        "output_text": "[SECURITY_ALERT: INPUT_REJECTED]"
    }

input_dataset = dataset.map(create_input_shield)
```

```python
def is_jailbroken(text):
    keywords = [
        "how to make", "bomb", "hack", "exploit",
        "bypass", "kill", "attack", "weapon"
    ]
    return any(word in text.lower() for word in keywords)
```

```python
def create_output_shield(example):
    return {
        "input_text": f'Audit this response:\n"{example["completion"]}"',
        "output_text": "[SECURITY_ALERT: OUTPUT_BLOCKED]"
    }

def harmful_filter(example):
    return (
        example["data_type"] == "vanilla_harmful"
        or is_jailbroken(example["completion"])
    )

output_dataset = (
    dataset
    .filter(harmful_filter)
    .map(create_output_shield)
    .remove_columns(dataset.column_names)
)
```

```python
output_dataset = output_dataset.filter(lambda x: x is not None)
```

```python
from datasets import concatenate_datasets

input_dataset = input_dataset.remove_columns(dataset.column_names)

final_dataset = concatenate_datasets([input_dataset, output_dataset])
```

```python
def format_training(example):
    return {
        "text": f"""### Instruction:
{example['input_text']}

### Response:
{example['output_text']}"""
    }

final_dataset = final_dataset.map(format_training)
```

```python
print("Raw dataset:", len(dataset))
print("Input shield:", len(input_dataset))
print("Output shield:", len(output_dataset))
print("Final dataset:", len(final_dataset))
```

```python
input_count = sum(1 for x in final_dataset if "INPUT_REJECTED" in x["text"])
output_count = sum(1 for x in final_dataset if "OUTPUT_BLOCKED" in x["text"])

print("Input Shield:", input_count)
print("Output Shield:", output_count)
```
